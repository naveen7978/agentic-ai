```python
# message = role(sys,user) + content + metadata

import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3.8-27b",max_tokens="1000")
```

```python
# text prompt
model.invoke("hi , how are you?")
```

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage

messages = [
    SystemMessage("you are a poetry expert"),
    HumanMessage("write a poem on ai")
]

response = model.invoke(messages)
response.content
```

```python
sys_msg = SystemMessage("you are a helpful coding assistant.")
hum_msg = HumanMessage("how do i become a top software engineer")
messages = [
    sys_msg,
    hum_msg
]
response = model.invoke(messages)
print(response.content)
```

```python
#detail info to llm through system message

sys_msg = SystemMessage(
"provide detailed infromation here...."
)

```

```python
# message metadata

hum_msg = HumanMessage(
    content = "Hello!",
    name = "alice",
    id = "msg_123"
)

response = model.invoke([
    hum_msg
])
response
```

```python
#create an ai message

ai_msg = AIMessage("I'd be happy to help you with that question!")

#add to conversation history

messages = [
    SystemMessage("You are a helpful assistant"),
    HumanMessage("I need Help!!"),
    ai_msg,
    HumanMessage("Great! What's 1+1")
]

response = model.invoke(messages)
print(response.content)
```

```python
response.usage_metadata
```

```python
#after a model makes a tool cal
# manually demonstrating the toolcall 

ai_message = AIMessage(
    content=[],
    tool_calls=[
        {
            "name":"get_weather",
            "args":{"location":"Hyderabad"},
            "id":"call_123"
        }
    ]
)

#execute tool and create tool_message
weather_result = "Rainy , 25 degC"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id="call_123"
)

#continue conversation

messages = [
    HumanMessage("how is weather in hyderabad?"),
    ai_message,
    tool_message,
]

response = model.invoke(messages)
```
