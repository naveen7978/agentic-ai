### binding tools with model

```python
from langchain.tools import tool

@tool
def get_weather(location:str)->str:
    """Fetch weather at a location"""
    return f"It is rainy in {location}"

model_with_tools = model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("how is weather in hyderabad?")
print(response)
for tool_call in response.tool_calls:
    #view tool calls made by the model
    print(tool_call)
    print(f"Tool : {tool_call['name']}")
    print(f"Args : {tool_call['args']}")
```

### tool execution loops

```python
# step 1 : model geenrates tool calls 
messages = [{"role":"user","content":"how is weather in hyderabad?"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

#step2 : execute tools and collect results
for tool_call in ai_msg.tool_calls:
    #execute the tool with the generated arguments
    tool_result = get_weather.invoke(tool_call)
    messages.append(tool_result)
    print(tool_result)
print(messages)
#step3 : pass results to model for the final response
final_res = model_with_tools.invoke(messages)
print(final_res.text)
    
```
